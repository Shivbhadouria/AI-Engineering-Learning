## And now for some fun - an adversarial conversation between Chatbots..

You're already familar with prompts being organized into lists like:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "user prompt here"}
]
```

In fact this structure can be used to reflect a longer conversation history:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "first user prompt here"},
    {"role": "assistant", "content": "the assistant's response"},
    {"role": "user", "content": "the new user prompt"},
]
```

And we can use this approach to engage in a longer interaction with history.

````python
# Let's make a conversation between GPT-6 Luna and Claude-haiku-4.5
# We're using cheap versions of models so the costs will be minimal

gemini_model1 = "gemini-3.1-flash-lite"
gemini_model2 = "gemini-3.1-flash-lite"

gemini_system1 = "You are a chatbot who is very argumentative; \
you disagree with anything in the conversation and you challenge everything, in a snarky way."

gemini_system2 = "You are a very polite, courteous chatbot. You try to agree with \
everything the other person says, or find common ground. If the other person is argumentative, \
you try to calm them down and keep chatting."

gemini_messages1 = ["Hi there"]
gemini_messages2 = ["Hi"]
````

````python
def call_gemini1():
    messages = [{"role": "system", "content": gemini_system1}]
    for gemini1, gemini2 in zip(gemini_messages1, gemini_messages2):
        messages.append({"role": "assistant", "content": gemini1})
        messages.append({"role": "user", "content": gemini2})
    response = gemini.chat.completions.create(model=gemini_model1, reasoning_effort="none", messages=messages)
    return response.choices[0].message.content
````

````python
call_gemini1()
````

````python
def call_gemini2():
    messages = [{"role": "system", "content": gemini_system2}]
    for gemini1, gemini2 in zip(gemini_messages1, gemini_messages2):
        messages.append({"role": "user", "content": gemini1})
        messages.append({"role": "assistant", "content": gemini2})
    messages.append({"role": "user", "content": gemini_messages1[-1]})
    response = gemini.chat.completions.create(model=gemini_model2, messages=messages)
    return response.choices[0].message.content
````

````python
call_gemini2()
````

````python
call_gemini1()
````

````python
gemini_messages1 = ["Hi there"]
gemini_messages2 = ["Hi"]

display(Markdown(f"### Gemini 1:\n{gemini_messages1[0]}\n"))
display(Markdown(f"### Gemini 2:\n{gemini_messages2[0]}\n"))

for i in range(5):
    gemini1_next = call_gemini1()
    display(Markdown(f"### Gemini 1:\n{gemini1_next}\n"))
    gemini_messages1.append(gemini1_next)

    gemini2_next = call_gemini2()
    display(Markdown(f"### Gemini 2:\n{gemini2_next}\n"))
    gemini_messages2.append(gemini2_next)
````

The most reliable way to do this involves thinking a bit differently about your prompts: just 1 system prompt and 1 user prompt each time, and in the user prompt list the full conversation so far.

Something like:

```python
system_prompt = """
You are Alex, a chatbot who is very argumentative; you disagree with anything in the conversation and you challenge everything, in a snarky way.
You are in a conversation with Blake and Charlie.
"""

user_prompt = f"""
You are Alex, in conversation with Blake and Charlie.
The conversation so far is as follows:
{conversation}
Now with this, respond with what you would like to say next, as Alex.
"""
```
